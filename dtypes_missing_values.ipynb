### notes 
1. Parameter `method` in `fillna()`  
   - `ffill` / `pad` —  fills in the gaps with the **previous value** (from top to bottom)  
   - `bfill` / `backfill` — fills in the gaps with the **following value** (from bottom to top)  
> It's also possible to use `bfill()` & `ffill()` methods. They do the same.

2. `replace()`
   Method `replace()` can be used for replacing things. E.g.:
   ```python
   import pandas as pd 

    s = pd.Series([1, 2, 3, 4, 5])
    
    # replacing one value 
    print(s.replace(2, 100).to_list())  # [1, 100, 3, 4, 5] 
    
    # replacing several values 
    print(s.replace([1, 2], 13).to_list())  # [13, 13, 3, 4, 5]
    print(s.replace({1: 13, 2: 1101}).to_list())  # [13, 1101, 3, 4, 5]
    
    # replacement in dataframe column 
    df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
    print(df['A'].replace(2, 99).to_list())  # [1, 99, 3]
    
    # replacement in all dataframe 
    df = pd.DataFrame({'A': [1, 2, 2], 'B': [2, 5, 6]})
    print(df.replace(2, 13))
    #     A   B
    # 0   1  13
    # 1  13   5
    # 2  13   6
    
    # replacement using regex 
    s = pd.Series(['cat', 'dog', 'cattle', 'doggy'])
    print(s.replace('^cat.*', 'animal', regex=True).to_list())  # ['animal', 'dog', 'animal', 'doggy']
```
